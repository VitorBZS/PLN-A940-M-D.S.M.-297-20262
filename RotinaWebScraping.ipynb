```python
import csv
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin
from google.colab import files

url = 'https://books.toscrape.com/'
livros = []

while url:
    pagina = requests.get(url)
    soup = BeautifulSoup(pagina.content, 'html.parser')

    for livro in soup.select('article.product_pod'):
        nome = livro.select_one('h3 a')['title']
        preco = livro.select_one('.price_color').get_text(strip=True)
        disponibilidade = livro.select_one('.availability').get_text(strip=True)
        livros.append([nome, preco, disponibilidade])

    proxima = soup.select_one('li.next a')
    url = urljoin(url, proxima['href']) if proxima else None

with open('livros.csv', 'w', newline='', encoding='utf-8-sig') as arquivo:
    escritor = csv.writer(arquivo, delimiter=';')
    escritor.writerow(['Nome', 'Preço', 'Disponibilidade'])
    escritor.writerows(livros)

print(f'Coleta concluída: {len(livros)} livros salvos em livros.csv')
files.download('livros.csv')
```
