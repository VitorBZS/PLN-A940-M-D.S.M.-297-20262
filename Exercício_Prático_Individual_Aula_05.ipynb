```python
!pip install -q nltk spacy pandas
!python -m spacy download pt_core_news_md -q

import nltk
import spacy
import pandas as pd

nltk.download('stopwords', quiet=True)
nltk.download('rslp', quiet=True)

nlp = spacy.load("pt_core_news_md")
```

##1. Tokenização

```python
frase_exercicio = "O proprietário alugou a casa para a família rapidamente."

doc = nlp(frase_exercicio)
tokens = [token.text for token in doc]
print("Tokens:", tokens)
```

##2. Remoção de pontuação e stopwors

```python
tokens_limpos = [token.text for token in doc if not token.is_punct and not token.is_stop]
print("Tokens Limpos:", tokens_limpos)
```

##3. Stemming e Lematização

```python
from nltk.stem import RSLPStemmer

stemmer = RSLPStemmer()
stems = [stemmer.stem(t) for t in tokens_limpos]

doc_limpo = nlp(" ".join(tokens_limpos))
lemas = [token.lemma_ for token in doc_limpo]

comparacao = pd.DataFrame({
    "Token" : tokens_limpos,
    "Stemming (RSLP)": stems,
    "Lematização (spaCy)": lemas
})

comparacao
```

##4. POS Tagging

```python
dados_pos = [{"Token": t.text, "POS": t.pos_ } for t in doc]
pd.DataFrame(dados_pos)
```

##5. Sintagma Nominal e Sintagma Verbal

```python
raiz = [token for token in doc if token.dep_ == "ROOT"][0]
sujeito = [token for token in doc if token.dep_ == "nsubj"][0]

sintagma_nominal = list(sujeito.subtree)
sintagma_verbal = [t for t in raiz.subtree if t not in sintagma_nominal]

print("Sintagma Nominal(SN):", [t.text for t in sintagma_nominal])
print("Sintagma Verbal (SV):", [t.text for t in sintagma_verbal])
```

##6. Similiariedade

```python
palavra_a = nlp("proprietário")
palavra_b = nlp("casa")

print(f"Similiariedade proprietário x casa: {palavra_a.similarity(palavra_b):.3f}")
```
