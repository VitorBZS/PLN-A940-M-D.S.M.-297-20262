```python
!pip install -q spacy nltk pandas
!python -m spacy download pt_core_news_sm -q

import spacy
import pandas as pd

nlp = spacy.load("pt_core_news_sm")
```

```python
frase_exercicio = "Olá! Gostaria de agendar uma visita para ver o imóvel de 3 quartos amanhã às 14h?"

# 1. Tokenização com spaCy
doc_spacy = nlp(frase_exercicio)
tokens_spacy = [token.text for token in doc_spacy]
print("Tokens com spaCy:", tokens_spacy)

# 2. Remoção de pontuação (use token.is_punct)
doc = nlp(frase_exercicio)
sem_pontuacao = []
for token in doc:
  if token.is_punct == False:
    sem_pontuacao.append(token.text)

print("Sem pontuação:", sem_pontuacao)


# 3. Contagem: original vs. sem pontuação
contagem_com_pontuacao = len(tokens_spacy)
print("\nContagem Original:", contagem_com_pontuacao)

contagem_sem_pontuacao = len(sem_pontuacao)
print("Contagem Sem Pontuação:", contagem_sem_pontuacao)


```
